```python
sbi = df[('Close', 'SBIN.NS')]
sbi_rtn = sbi.pct_change()
sbi_rtn
```

```python
sbi_rtn=sbi_rtn.dropna()
sbi_rtn
```

```python
df_rolling=sbi_rtn.rolling(window=21).agg(["mean","std"])
df_rolling
```

```python
df_outliers = pd.concat([sbi_rtn.rename('sbi_rtn'), df_rolling], axis=1)
#df_outliers = df_outliers.dropna()
```

```python
def identify_outliers(row,n_sigma=3):
  x=row["sbi_rtn"]
  mu=row["mean"]
  sigma=row["std"]
  if(x>mu+3*sigma)|(x<mu-3*sigma):
    return 1
  else:
    return 0
df_outliers["outlier"] = df_outliers.apply(identify_outliers, axis=1)
df_outliers.head()
```
